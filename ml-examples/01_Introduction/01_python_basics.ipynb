# Introduction to Python Basics


This script covers some fundamental Python concepts for beginners.

You can use the ```print``` command to show some text (or variables, etc.):

````python
print('Hello, world!')  # This will show the text "Hello, world!"
print()                 # This will just show an empty line
print(5)                # This will show the number 5
print('5')              # This will show the text '5'
print('5' * 10)        # This will show the text '5' repeated 10 times

# We can add comments to our code using the hash ('#') character.
# Anything after the # on that line is ignored by Python.
````

````python
print('We can show text in single quotes.')       # This will work
print("We can also show text in double quotes.")  # This will work too

# print("Quotes need to match')  # This will cause an error
````

## 1. Variables

````python
# Strings
name = "Alice"               # 'Strings' are text enclosed in quotes ( ' or " )
greeting = "Hello, " + name  # Can combine strings using a plus sign
print(greeting)

# Another way to combine:
greeting = f"Hello, {name}"  # Can combine strings using an f-string (note the f before the quotes)
print(greeting)
````

````python
print()  # Shows an empty line

# Integer Numbers, i.e. whole numbers
integer_num = 42

print(f'Integer: {integer_num}')       # Shows an integer number
print('Integer: ' + str(integer_num))  # Another way to show an integer
# print('Integer: ' + integer_num)     # This will cause an error (cannot add text to a number!)

print()  # Shows an empty line

print(f'Add:       42 + 1  = {integer_num + 1}')  # Shows the integer plus one
print(f'Subtract:  42 - 1  = {integer_num - 1}')  # Shows the integer minus one
print(f'Multiply:  42 * 2  = {integer_num * 2}')  # Shows the integer times two
print(f'Divide:    42 / 2  = {integer_num / 2}')  # Shows the integer divided by two
print(f'Remainder: 42 % 3  = {integer_num % 3}')  # Shows the remainder when the integer is divided by 3
print(f'Squared:   42 ** 2 = {integer_num ** 2}') # Shows the integer squared
print(f'Cubed:     42 ** 3 = {integer_num ** 3}') # Shows the integer cubed

print()  # Shows an empty line
````

````python
# Floating Point Numbers, i.e. numbers with decimal points

float_num = 3.14
print(f'Float: {float_num}')  # A 'float' is a number with a decimal point
print()

print(f'Add one: {float_num + 1.0}')  # Shows the float plus one
print()

print(f'Round to 2 places: {round(3.123456, 2)}')  # Rounds to 2 decimal places
print()

# Can add integers and floats together
print(f'Integer + Float: {integer_num} + {float_num} = {integer_num + float_num}')
print()
````

````python
# Boolean, i.e. True or False
is_student = True
print(f"Is student: {is_student}")  # A 'boolean' is either True or False
print(f"Is not student: {not is_student}")  # The 'not' operator negates the boolean value
````

````python
# Show the type of a variable
print(type(5))        # int   = integer number
print(type(3.14))     # float = decimal number
print(type("Hello"))  # str   = string (i.e. text)
print(type(True))     # bool  = boolean (i.e. True/False)
````

## 2. Control Flow (if statements)

````python
print("'if-else' example:")
print()  # Shows an empty line

age = 25  # You can change this value to test different outputs

# Guide:
# ==    means: 'equal to'
# >     means: 'greater than'
# >=    means: 'greater than or equal to'
# <     means: 'less than'
# <=    means: 'less than or equal to'


if age >= 18:  # If age is 18 or older, then:
    status = "Is an adult"

else:  # Otherwise:
    status = "Is not an adult"

print(age)
print(status)
````

````python
print("Example including elif:")
print()

age = 85

if age >= 68:  # If age is 68 or older, then:
    status = "This is a retired adult"

elif age == 67:  # If age is exactly 67, then:
    status = "This is an adult about to retire"

elif age >= 18:  # Otherwise, if age is 18 or older, then:
    status = "This is an adult"

else:  # Otherwise:
    status = "This is a minor"

print(f"Age: {age}, Status: {status}")
````

## 3. Lists (ordered, mutable collection)

````python
print("Lists:")
print()  # Shows an empty line

numbers = [1, 2, 3, 4, 5]
print(f"Original list:  {numbers}")

# Can get a value from the list by position (counting starts at zero)
# The position of a value is called its 'index'
print(f"First element:  {numbers[0]}")
print(f"Second element: {numbers[1]}")
print()

# Can add values to the end of the list
numbers.append(6)
print(f"Add 6 to the list: {numbers} \n")  # Note: '\n' adds an empty line

# Can remove a value from the list by position
numbers.pop(0)
print(f"Removed first value: {numbers} \n")

# Can get the length of the list
print(f"Length: {len(numbers)} \n")

# Sum of all values in the list
print(f"Sum: {sum(numbers)} \n")

# Average of all values in the list
average = sum(numbers) / len(numbers)  # sum divided by length
print(f"Average: {average} \n")
````

## 4. Dictionaries

````python
print("Dictionaries:")
print()

# A 'dictionary' is a collection of key-value pairs (enclosed in curly braces)
# Here, we create a dictionary to represent a student.
# The student has a name, age, and a list of grades.
student = {
    "name": "Bob",          # This is a string
    "age": 20,              # This is an integer
    "grades": [85, 90, 92]  # This is a list
}

print(f"Student: {student}")  # Print the whole dictionary

print(f"Name: {student['name']}")  # Print the name only

print(f"Grades: {student['grades']}") # Print the grades only
````

## 5. Loops

````python
print("Example of a 'for' loop")

print("Counting from 0 to 5:")

for num in range(6):  # range(6) generates numbers from 0 to 5
    print(num)        # Print each number. Note the spaces
````

````python
print("Example of a 'while' loop")

count = 0  # Start at 0

print("Counting from 0 to 5:")
while count < 6:  # Continue while count is less than 6

    print(f"  {count}")  # Print the current count

    count = count + 1    # Increase count by 1

print()  # Shows an empty line

count = 3
print("Counting from 3 to 1:")
while count >= 1:        # Continue while count is 1 or more
    print(f"  {count}")  # Print the current count
    count = count - 1    # Decrease count by 1

````

## 6. Functions

````python
print("Functions \n")

# Functions are a way to group code together to perform a specific task.
# Functions are defined using the 'def' keyword

# This function states whether a person is an adult based on their age
def check_adult(age):
    if age >= 18:
        print("Is an adult")
    else:
        print("Is not an adult")

check_adult(20)  # Call the function with age 20
check_adult(15)  # Call the function with age 15
check_adult(100)  # Call the function with age 100

print()  # Shows an empty line
````
