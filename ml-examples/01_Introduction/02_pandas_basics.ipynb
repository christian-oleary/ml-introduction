# Introduction to the pandas package

In Python, a package is a compilation of functions that you can use

This script some basic *pandas* code.

The *pandas* package can be used to load and handle data.

### Preparing some data (ignore this section for now)

```python
import pandas as pd  # Tell Python to use the pandas package and call it 'pd' for short
```

```python
# Create some example data
# (Don't worry about the details here, we'll cover this later! Just run the code for now)

# Create some example data (as a dictionary)
some_example_data = {
    'Name':  ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age':   [25, 20, 35, 28, 42],
    'Score': [85, 92, 78, 88, 95],
    'City':  ['Paris', 'London', 'Paris', 'Berlin', 'Berlin'],
}

# This saves the data to a Excel file named 'students.csv' (will explain later!)
pd.DataFrame(some_example_data).to_csv('students.csv', index=False)

print("Example data saved to 'students.csv'\n")
```

```python
# A short function to print a separator line
def blank_line():
    print("_" * 30 + "\n")
```

## 1. Creating DataFrames

```python
blank_line()
print("Creating DataFrames")
blank_line()

# A 'DataFrame' is a table of data with rows and columns.

# Load the data from the 'students.csv' file into a DataFrame
df = pd.read_csv('students.csv')

print("Student DataFrame:\n")

print(df)  # Notice how the DataFrame shows the data in a nice table format.

blank_line()  # A border and a blank line
```

## 2. Viewing Data

```python
print("Show the first 3 rows: \n")
blank_line()  # Show a border and a blank line

print(df.head(3))  # Show the first 3 rows of the DataFrame

blank_line()
```

```python
print("Show the column names: \n")
blank_line()

print(df.columns.tolist())  # Show the column names (as a list)

blank_line()
```

## 3. Selecting Data

```python
print("Show the 'Names' column:")
blank_line()

print(df['Name'])  # Select the 'Name' column

blank_line()
```

```python
# Select multiple columns
print("Show the 'Names' and 'Scores' columns:")
blank_line()

print(df[['Name', 'Score']])

blank_line()
```

```python
# Select rows by position
print("Show the first row:")
print(df.iloc[0])  # First row = 0

blank_line()

# Show the second row:
print("Show the second row:")
print(df.iloc[1])  # Second row = 1

blank_line()
```

```python
blank_line()

# Select rows by condition
print("Students with Age > 30:")
print(df[df['Age'] > 30])

blank_line()

# Select rows with 'Age' less than 30
print("Students with Age < 30:")
print(df[df['Age'] < 30])

blank_line()
```

## 4. Adding and Modifying Columns

```python
print("\n Adding a new column called 'is_student': \n")

# Add a new column called 'is_student' and set all values to True
df['is_student'] = True

print(df)
```

```python
print("\n Adding a 'Pass' column: \n")

# Add a new column called 'Pass': True if Score >= 50, else False
df['Pass'] = df['Score'] >= 50

print(df)
```

## 5. Filtering and Sorting

```python
print("\n High scorers (Score >= 90): \n")

high_scorers = df[df['Score'] >= 90]  # Filter rows where Score >= 90

print(high_scorers[['Name', 'Score']])  # Show only 'Name' and 'Score' columns of high_scorers
```

```python
# Multiple conditions
print("\nHigh scorers under 30:")

young_high_scorers = df[(df['Score'] >= 90) & (df['Age'] < 30)]

print(young_high_scorers[['Name', 'Age', 'Score']])
```

## 6. Statistics

```python
# Basic statistics
print(f"Mean Score: {df['Score'].mean()}")
print(f"Max Score:  {df['Score'].max()}")
print(f"Min Score:  {df['Score'].min()}")

blank_line()
```

```python
# Count values
print("Pass/Fail counts:")
blank_line()

counts = df['Pass'].value_counts()

print(counts)
```

## 7. Reading and Writing Files

```python
# Save to CSV
csv_file = 'students.csv'

df.to_csv(csv_file, index=False)  # index=False to avoid saving the row numbers

print(f"Saved DataFrame to {csv_file}")
```

```python
# Read from CSV
df_new = pd.read_csv(csv_file)

print("Loaded DataFrame:")
blank_line()
print(df_new.head())
```
